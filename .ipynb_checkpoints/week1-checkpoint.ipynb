```python
basic = int (input("Enter your Basic salary"))

if(basic<=10000):
    hra=0.67* basic
    da=0.8*basic
elif(basic<=20000):
    hra=0.69*basic
    da=0.82*basic
else:
    hra=0.72*basic
    da=0.90*basic

salary=hra+da+basic
print("Total gross salary is =",salary)
    
```

```python
gross=int(input())
tax=0
if(gross<=700000):
    print("NO TAX NEEDED")
else:
    gross-=300000
    if(gross>0):
        cut=min(gross,300000)
        tax+=0.05*cut
        gross-=cut
    if(gross>0):
        cut=min(gross,600000)
        tax+=0.15*cut
        gross-=600000
    if(gross>0):
        cut=min(gross,600000)
        tax+=0.30*gross
        
print("You TDS is:" ,tax)
```

### String Methods

```python
statement="    Python is a , programming Language    "

# Removing Spaces using differnt methods
print(statement.strip());
print(statement.rstrip())
print(statement.lstrip())
print(statement.strip())
```

```python
statement="    Python programming Language    "

# Convet into lower and upper case
print(statement.lower().strip())
print(statement.upper().strip())

# Check if the statement is lower or upper
print(statement.islower())
print(statement.isupper())

# Title Method
print(statement.title())
print(statement.istitle())
```

```python
word="swiss"
for i in word:
    if(word.count(i)==1):
        print(i)
        break
```

```python
from collections import Counter

count=Counter(word)
print(count)
for i in word:
    if(count[i]==1):
        print(i)
        break
```

### Task

```python
amount=int(input("Enter amount to Withdraw"))
notes=0;
if(amount%100==0):
    if(amount !=0 and amount//500>0):
        note=amount//500
        notes+=note
        print("Number of 500 Notes are:",note)
        amount-=(500*note)
    if(amount !=0 and amount//200>0):
        note=amount//200
        notes+=note
        print("Number of 200 Notes are:",note)
        amount-=(200*note)
    if(amount !=0 and amount//100>0):
        note=amount//100
        notes+=note
        print("Number of 100 Notes are:",note)
        amount-=(100*note)
    print("TOtal number of notes :",notes);
else:
    print("Please enter Valid Amount, Must be multiple or 100")
```
