```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
import time
```

```python
# importing URL of the website
url='https://www.amazon.in/s?crid=2QMOFAFMZVWHB&i=aps&k=laptop&ref=nb_sb_noss_1&sprefix=laptop%2Caps%2C446&url=search-alias%3Daps'
```

```python
# Create the empty list to store te data
headers={
    "User-Agent":"Mozilla/5.0 (Linux; Android 6.0; Nexus 5 Build/MRA58N) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Mobile Safari/537.36"
}
```

```python
data=[]
```

```python
#scrape the data from the website
response =requests.get(url,headers=headers)
#check status code
response.status_code
```

```python
# scrape the data from the website

for page in range(1,3):
    params={'k':'laptop','page':page}
    # make the request with parameters
    response=requests.get(url,headers=headers,params=params)
response.status_code
```

```python
# Run this command to watch entire content in html format <CAUTION  : Very long response>
# response.content
```

```python

for page in range(1,2):
    params={'k':'laptop','page':page}
    # make the request with parameters
    response=requests.get(url,headers=headers,params=params)

    soup=BeautifulSoup(response.content,'html.parser')

    #Find all the products on the page
    products=soup.find_all('div',{'data-component-type':'s-search-result'})
    
    #Extract all the data
    for product in products:
        
        title_tag=product.find("h2")
        if not title_tag:
            continue
        title_text=title_tag.get_text(strip=True)
        
        price_tag=product.find("span",{"class":"a-price-whole"})
        if not price_tag:
            continue
        price_text=price_tag.get_text(strip=True)

        match=re.match(r'^\W*([A-Za-z]+)',title_text)
        brand=match.group(1).upper() if match else 'UNKNOWN'

        
        rating_tag = product.find("span", {"class": "a-icon-alt"})
        if rating_tag:
            rating_match = re.search(r'(\d+\.?\d*)', rating_tag.get_text())
            rating = rating_match.group(1) if rating_match else "N/A"
        else:
            rating = "N/A"

        # color_tag = product.find("span", class_="a-size-base")
        # if color_tag:
        #     color_match = re.search(r'\s*([A-Za-z\s]+)', color_tag.get_text())
        #     color = color_match.group(1).strip() if color_match else "N/A"

        colors = [
            "Rose Gold", "Space Gray", "Midnight Black",
            "Black", "White", "Silver", "Gray", "Grey",
            "Blue", "Red", "Green", "Yellow", "Pink", "Purple", "Gold"
        ]
        
        color_pattern = r'\b(' + '|'.join(map(re.escape, colors)) + r')\b'
        color_match = re.search(color_pattern, title_text, re.IGNORECASE)
        
        color = color_match.group(1) if color_match else "N/A"
        
        data.append({
            'Title':title_text,
            'Price':price_text,
            'Brand':brand,
            'Color':color,
            'Rating':rating,
        })
    print(f"page {page} scraped")
    time.sleep(1)
    
```

```python
for product in data:
    print("Title:",product['Title'])
    print("Price:",product['Price'])
    print("Brand:",product['Brand'])
    print("Color:",product['Color'])
    
    print("Rating",product['Rating'])
    
    print('-'*50)
    
```
