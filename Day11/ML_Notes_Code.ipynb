# Linear Regression Model


```python
# step 1: Import the required lib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# step 2: Load the datasets -> head, tail, info, columns

df=pd.read_csv('linear_regression_datasest.csv')
df
```

```python
# step 3: Data preprocessing -> missing , detect outlier

df.describe()
```

```python
# step 4: EDA ( Expolratory Data Analysis)
# step 5: Feature Engineering

plt.figure(figsize=(10,6))

sns .boxplot(data=df,color="lavender")
plt.title("Outlier Detection")
plt.xlabel("Features")
plt.show()
```

```python
sns.scatterplot(x='Feature_2',y='Target',data=df)
plt.show()
```

```python
# step 6: Divide data into x and y (input features and output features)

x=df.drop('Target',axis=1)
y=df['Target']
```

```python
x.shape, y.shape
```

```python
# step 7: Splitting dataset into training and testing  

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# step 8: Train model 

# 1. Import model LinearRegression
from sklearn.linear_model import LinearRegression

# 2. Load model LinearRegression
model=LinearRegression()

# 3. Fit model on the trained data(ie: train model)
model.fit(x_train,y_train)

# 4. Predict the testing data 
y_pred=model.predict(x_test)
y_pred
```

```python
results=pd.DataFrame({'Actual':y_test,'Predicted':y_pred})
results
```

```python
from sklearn.metrics import mean_squared_error

mse=mean_squared_error(y_test,y_pred)
print("Mean Squared Error",mse)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
print("R2 Score:",r2)
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(x=y_test,y=y_pred,color='teal')
plt.plot([y.min(),y.max()],[y.min(),y.max()],'r--')
plt.title("Actualvs Predicted Values")
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.show()
```
