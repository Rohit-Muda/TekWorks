```python
import pandas as pd
df = pd.read_csv("flight_delay_linear_regression_dataset.csv")
df
```

```python
import matplotlib.pyplot as plt
import numpy as np;

plt.scatter(df['distance_km'],df['arrival_delay_min'])
m,c=np.polyfit(df['distance_km'],df['arrival_delay_min'],1)
plt.plot(df['distance_km'],m*df['distance_km']+c)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X = df['distance_km']
y = df["arrival_delay_min"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python

model = LinearRegression()

X=X.to_numpy().reshape(-1,1)
model.fit(X,y)


predictions = model.predict([[3360]])
predictions
```
