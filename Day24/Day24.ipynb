# Ensemble Learning

## Introduction
Ensemble Learning is a Machine Learning technique where multiple models are combined to improve overall performance.

Instead of relying on one model:
- Multiple models work together
- Accuracy increases
- Overfitting reduces
- Predictions become more stable

---

# Types of Ensemble Learning

## 1. Bagging (Bootstrap Aggregating)
- Models are trained in parallel
- Each model gets random subsets of data
- Final output is based on voting/averaging

### Example:
- Random Forest

---

## 2. Boosting
- Models are trained sequentially
- Each new model focuses on previous model mistakes
- Weak learners become strong learners

### Examples:
- AdaBoost
- Gradient Boosting
- XGBoost

---

# Bagging

## Random Forest
Random Forest is a bagging technique used to overcome overfitting caused by Decision Trees.

### Working:
- Multiple decision trees are created
- Each tree gets random data samples
- Trees work in parallel
- Final prediction is based on majority voting

### Advantages:
- Reduces overfitting
- Better accuracy
- Stable model

---

# Boosting

## Concept
Boosting improves weak learners step by step.

### Key Idea:
- Identify weak points/errors
- Increase focus on wrongly classified data
- Train next model on those errors

### Important:
- Boosting works sequentially
- Random Forest works in parallel

---

# AdaBoost (Adaptive Boosting)

AdaBoost mainly uses:
- Decision Trees
- Mostly Decision Stumps (depth = 1)

---

## AdaBoost Formula

\[
F = \alpha_1(m_1) + \alpha_2(m_2) + \alpha_3(m_3) + ... + \alpha_n(m_n)
\]

Where:
- \(m\) = model/stump
- \(\alpha\) = weight/performance of stump

---

# Performance of Stump

\[
Performance = \frac{1}{2} \times \ln \left(\frac{1-TE}{TE}\right)
\]

Where:
- TE = Total Error

---

# Steps in AdaBoost

## Step 1: Assign Initial Weights
- Equal weights are assigned to all data points.

---

## Step 2: Train First Stump
- Build a decision stump.
- Calculate classification error.

---

## Step 3: Compute Performance of Stump

\[
Performance = \frac{1}{2} \times \ln \left(\frac{1-TE}{TE}\right)
\]

Higher performance:
- Better stump

Lower performance:
- Weak stump

---

## Step 4: Update Weights

### For Correctly Classified Points

\[
Weight \times e^{-Performance}
\]

### For Incorrectly Classified Points

\[
Weight \times e^{Performance}
\]

Meaning:
- Incorrect points get higher weight
- Correct points get lower weight

So next stump focuses more on mistakes.

---

## Step 5: Normalize Weights
- Convert weights into probabilities.
- Sum of all weights becomes 1.

---

## Step 6: Create Bins
- Weight ranges are converted into bins.

---

## Step 7: Select Data for Next Stump
- Random values between 0 and 1 are generated.
- Based on bins, data points are selected.

### Important:
- Error-prone records have higher probability of selection.

---

## Step 8: Train Next Stump
- New stump focuses more on previously misclassified points.

This process repeats sequentially.

---

# Gradient Boosting
- Improves errors using gradient descent concept
- Each model tries to reduce residual errors

---

# XGBoost

Advanced version of Gradient Boosting.

## Features:
- Faster
- Regularization support
- Parallel optimization
- High accuracy
- Widely used in competitions

---

# Difference Between Bagging and Boosting

| Feature | Bagging | Boosting |
|---|---|---|
| Training | Parallel | Sequential |
| Goal | Reduce overfitting | Reduce errors |
| Model Dependency | Independent | Dependent |
| Example | Random Forest | AdaBoost, XGBoost |
| Focus | Variance reduction | Bias reduction |
