## Create the Database in MySQL

```python
cursor= conn.cursor()                   #cursor() method is used to fill the gap between the database and the python code(application)
cursor.execute("CREATE DATABASE IF NOT EXISTS Students_db")  #execute() method is used to execute the given database operation(query)
print("Database created successfully")

```

## Connect to the database

```python
conn=connector.connect(
    host="localhost",
    user="root",
    password="Mudarohit@2005",
    database="Students_db"
    )
cursor= conn.cursor()
print("Connected to database : ",conn.is_connected())
```

## Step 4 Create Table

```python
cursor.execute("CREATE TABLE IF NOT EXISTS Students(id INT AUTO_INCREMENT PRIMARY KEY, name VARCHAR(255), age INT)")
print("Table created successfully")

```

## Step 5 Insert values into table

```python
cursor.execute("INSERT INTO Students (name,age) VALUES ('Rohit',19)")
conn.commit()
print("Record inserted successfully")
```

```python
sql="INSERT INTO Students (id,name,age) VALUES (%s,%s,%s)"

values=[
    (3,'Alice',20),
    (4,'Bob',21),
    (5,'Charlie',22)
]
cursor.executemany(sql,values)
conn.commit()
print("Multiple records have been inserterd")
```

## Display Table Data

```python
cursor.execute("SELECT * FROM Students")
records=cursor.fetchall()

for row in records:
    print(row)
```

## Delete Table Rows

```python
cursor.execute("DELETE FROM Students WHERE id=2 and name ='Rohit' ")
conn.commit()
print("Record deleted successfully")
```

## Update Table Rows

```python
cursor.execute("UPDATE Students SET id=4 WHERE id=5")
conn.commit()
```
