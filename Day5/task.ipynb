### 4

```python
import numpy as np
df['user_id']=df['user_id'].replace('', np.nan).str.upper()

df.sample(10)
```

## 5

```python
df['age'] = pd.to_numeric(df['age'],errors="coerce")
df.loc[(df['age']<0) | (df["age"]>100) ,"age"]=np.nan

df['age']
```

## 6

```python
df['salary']=df['salary'].astype(str).str.replace("₹","",regex="False").replace("N/A",np.nan)
df['salary']= pd.to_numeric(df['salary'],errors="coerce")
df.sample(10)
```

## 7

```python
df['join_date']=pd.to_datetime(df['join_date'],errors="coerce",dayfirst=True)

df['join_date']
```

```python
df.head(10)
```

## 8

```python
df['joining_year']=df['join_date'].dt.year
df['joining_year']
```

## 9

```python
df['department']=df['department'].str.strip().str.upper()
df['department']
```

## 10

```python
df['city']=df['city'].str.upper()
df['city']
```

## 11

```python
df['performance_score']=df['performance_score'].replace({"excellent": 5,"poor":0})
df['performance_score']=pd.to_numeric(df['performance_score'],errors="coerce")

df.sample(10)
```

## 12

```python
df['experience_years']=df['experience_years'].replace({"five":5})
df['experience_years']=pd.to_numeric(df['experience_years'],errors="coerce")
df.loc[(df['experience_years']<0) ,"experience_years"]=np.nan

df['experience_years']
```

## 13

```python
df['is_active']=df['is_active'].replace({"yes":True,"no":False})
df['is_active']=df['is_active'].astype("boolean")
df['is_active']
```

## 14

```python
df['user_id'].drop_duplicates(inplace=True)
df['user_id']
```
