## 

```python
df['order_date']=pd.to_datetime(df['order_date'],errors="coerce",dayfirst=True)
df
```

```python
import numpy as np
df['delivery_days']=pd.to_numeric(df['delivery_days'],errors="coerce")
# df.loc[(df['delivery_days']<0),'delivery_days']=np.nan

df[(df['order_date'].notna()) & (df['delivery_days']<0)]
```

```python
df['city']=df['city'].astype(str).str.strip().str.title()
df['payment_mode']=df['payment_mode'].astype(str).str.strip().str.lower()

x=(
    df.groupby(['city','payment_mode']).size()
)
perc=x/x.groupby(level=0).sum()*100
# perc=pd.DataFrame(perc,columns=(['percentage'])).reset_index()
perc=perc.reset_index(name='percentage')

perc
```

```python
df['customer_age']=pd.to_numeric(df['customer_age'],errors='coerce')
df['order_amount']=pd.to_numeric(df['order_amount'],errors='coerce')
x=df['order_amount'].median()
x
data=df[(df['customer_age'].isna()) & (df['order_amount']>x)]
data
```

```python
ucity=df.groupby('customer_id')['city'].nunique()

cust_order = ucity[(ucity>=1)].reset_index()
cust_order
```

```python
flag=df[ (df['delivery_days'].isna()) & (df['is_delivered']=='True')]
flag
```

```python
df['delivery_flag']=(df['delivery_days'].isna()) & (df['is_delivered'].astype(str).str.lower().isin(['True','true','yes']))
df['delivery_flag'].value_counts()
```

```python
city_missing=df.assign(missing=df['order_date'].isna()).groupby('city')['missing'].mean().mul(100)
city_missing[city_missing>30]
```
