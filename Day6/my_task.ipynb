```python
import pandas as pd
```

```python
df=pd.read_csv("Dataset_1.csv")
df.sample(10)
```

```python
df['order_amount']= pd.to_numeric(df['order_amount'],errors="coerce")
df['order_amount'].isna()
```

```python
df['is_delivered']=df['is_delivered'].replace({'yes':'true','no':'false','True':'true','False':'false'})
df.sample(10)
```

```python
import numpy as np
df['delivery_days']=pd.to_numeric(df['delivery_days'],errors="coerce")
df.loc[(df['delivery_days']<0),'delivery_days']=np.nan
```

```python
df['city']=df['city'].astype(str).str.title()
df['payment_mode']=df['payment_mode'].astype(str).str.lower()
cc=df.groupby('city')[['payment_mode']].value_counts(normalize=True)*100
cc
```

```python
df['city']=df['city'].astype(str).str.strip().str.title()
df['payment_mode']=df['payment_mode'].astype(str).str.strip().str.lower()
cc=df.groupby('city')[['payment_mode','city']].value_counts(normalize=True)*100
cc
```

```python
df['customer_age']=pd.to_numeric(df['customer_age'],errors='coerce')
df['order_amount']=pd.to_numeric(df['order_amount'],errors='coerce')
x=df['order_amount'].median()
x
df[(df['customer_age'].isna()) & (df['order_amount']>x)]
```

```python
ucity=df.groupby('customer_id')['city'].nunique()

cust_order = ucity[(ucity>=1)].reset_index()
cust_order
```

```python
flag=df[ (df['delivery_days'].isna()) & (df['is_delivered']=='true')]
flag
```

```python
df['delivery_flag']=(df['delivery_days'].isna()) & (df['is_delivered'].astype(str).str.lower().isin(['True','true','yes']))
df['delivery_flag']
df[['delivery_days','is_delivered','delivery_flag']]
```

```python
df['payment_mode']=df['payment_mode'].astype(str).str.strip().str.lower()
df.groupby('payment_mode')['order_amount'].mean().reset_index(name="Average_amount")
```
