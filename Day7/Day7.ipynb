```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
```

```python
URL="https://www.amazon.in/s?k=laptop&crid=N5G6QQN7BELB&sprefix=laptop%2Caps%2C402&ref=nb_sb_noss_2"
```

# Make header to mimic browser visit

```python
#make  header to mimic a browser header
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36"
}
```

```python
# send GET request to the URL 
response=requests.get(URL,headers=headers)
response
```

```python
# beautifulSoup to parse the HTML content
soup=BeautifulSoup(response.content,"html.parser")
soup
```

```python
products=soup.find_all("div",class_="sg-col-inner")
len(products)
```

```python
product_names=[]
for product in products:
    title_tag=product.h2
    if title_tag:
        product_name=title_tag.text.strip()
        product_names.append(product_name)
    price_tag=products.find("span",class_="a-price-whole")
    if price_tag:
        product_price=price_tag.text.strip()
        product_names.append(product_price)
    else:
        product_names.append("N/A")
    
```
