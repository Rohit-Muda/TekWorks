```python
import numpy as np
import pandas as pd
df=pd.read_csv("transactions.csv")
df
```

```python
#What is the total sales amount across all transactions?
x=pd.to_numeric(df['t_amt'],errors='coerce').sum()
x
```

```python
#Which month had the highest total transaction amount?
df['t_date']=pd.to_datetime(df['t_date'],dayfirst=True,errors="coerce")
x=df.groupby(df['t_date'].dt.month)['t_amt'].sum()
x.columns = ['Month', 'TotalTransactionValue']
x
```

```python
#What is the average transaction amount per customer?
df.groupby('cust_id')['t_amt'].mean()
```

```python
#What is the trend of total sales over months?
#total sales done in each month in terms of total transactions done 
df.groupby(df['t_date'].dt.month)['t_amt'].sum()
```

```python
#total sales done in each month in terms of number of customers
df.groupby(df['t_date'].dt.month)['cust_id'].count()
```

```python
#What is the highest single transaction amount recorded?
df['t_amt'].max()
```

```python
#Which service category contributes the most to total revenue?
df.groupby('services')['t_amt'].sum()
```

```python
#Which product generated the highest revenue?
df.groupby('products_used')['t_amt'].sum().idxmax()
```

```python
#How does the average transaction amount vary between services?
df.groupby('services')['t_amt'].mean()
```

```python
#How many unique customers made purchases?
df['cust_id'].nunique()
```

```python
#Which customers have spent the most overall?
df.groupby('cust_id')['t_amt'].sum()
```

```python
#What is the average number of transactions per customer?
df.groupby('cust_id')['t_amt'].mean()
```

```python
#Are there customers who made multiple purchases in different categories?
df.groupby('cust_id')['services'].value_counts()
```

```python
#What percentage of customers are repeat buyers?
x=(df['cust_id'].count())-(df['cust_id'].nunique())
x=(x/(df['cust_id'].count()))*100
x
```

```python
#Compare average spending between California and Texas customers.
cal=df.loc[(df["state"]=="California")].groupby('state')['t_amt'].sum()
tex=df.loc[(df["state"]=="Texas")].groupby('state')['t_amt'].sum()
avg=(float(cal)+float(tex))/2
avg
```
