```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('dataset.csv')
df
```

# 1 Histogram

```python
# figure size 
plt.figure(figsize=(10,6))         # set the figure size

#histogram
sns.histplot(df['age'],kde=True,bins=30,color='red',edgecolor='black',alpha=0.7)

#set title and axis names of x and y
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

#set x-axis and y-axis limit
plt.xlim(25,80)
plt.ylim(0,35)


plt.show()
```

# 2. Scatter Plot

### Scatter plot uses:
1. Coorelation
2. 

```python
#figure size
plt.figure(figsize=(10,6))

#scatter plot
sns.scatterplot(x='age',y='chol',data=df,alpha=0.7,color="teal",edgecolor="black",s=100,marker="*")
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='trestbps',y='chol',data=df,alpha=0.7,color="pink",edgecolor="red",s=100,marker='$♥$')
```

# 3. Line Chart

```python
plt.figure(figsize=(10,6))

#Line plot
sns.lineplot(x='age',y='chol',data=df,alpha=1,color="teal",marker='$♥$')

plt.title("Age VS Chol")
```

# 4. Box Plot

### Uses:
1. outliers
2. range

```python

plt.figure(figsize=(10,6))
sns.boxplot(y='chol',x='age',data=df,color='gold')
plt.title("Cholestrol vs You")
plt.gca().set_facecolor("black")
```

```python

plt.figure(figsize=(6,10))
sns.boxplot(y='age',data=df,color='teal')
```

# 5 . Pie Chart

```python


plt.figure(figsize=(6,6))
plt.pie(
    df['sex'].value_counts(),
    labels=['🧔 Male', '👩 Female'],
    autopct='%1.2f%%',
    startangle=180,
    colors=['#00BFFF', '#FF69B4'],
    explode=(0.08, 0.02),
    shadow=True,
    pctdistance=0.75,
    labeldistance=1.1,
    counterclock=False,
    wedgeprops={
        'edgecolor': 'yellow',
        'linewidth': 2,
        'linestyle': 'solid'
    },
    textprops={
        'fontsize': 12,
        'weight': 'bold',
        'color': 'yellow'
    }
)

plt.title("Distribution of Target Variable")
plt.show()
```

# 6. Bar Chart

```python
plt.figure(figsize=(5,5))

sns.barplot(x='sex',y='age',data=df,palette=['skyblue', 'hotpink'])
plt.title("🕺 Who's Older on Average? 💃", fontsize=14)
plt.xlabel("Team")
plt.ylabel("Average Age")
```
